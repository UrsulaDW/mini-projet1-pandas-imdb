## Créer une colonne indiquant si un film est : Court (< 90 min) Moyen (90-120 min) Long (> 120 min)

```python
def timing(time) :
    if time < 90 :
        return "Court"
    elif time <= 120 :
        return "Moyen"
    else :
        return "Long"
    
movies['Timing'] = movies.Runtime.apply(timing)
movies.Timing 
```

## Créer une colonne "Décennie".

```python
def convert_decade(year) :
    x = year // 10
    x = x * 10
    return x

movies['Decade'] = movies.Released_Year.apply(convert_decade)
movies.Decade
```

## Créer une colonne indiquant si un film est "Très bien noté" (IMDb ≥ 8).

```python
movies['Is_Good'] = movies.IMDB_Rating.apply(lambda x : x>=8)
movies.Is_Good
```

## Les critiques (Metascore) sont-elles généralement plus sévères que les utilisateurs (IMDb) ?

```python
moyenne_Meta_score = movies.Meta_score.mean()
moyenne_IMDB_Rating = movies.IMDB_Rating.mean() * 10

if(moyenne_Meta_score > moyenne_IMDB_Rating) :
    print("Oui les critiques Metascore sont plus séveres que les utilisateurs IMDB")
elif(moyenne_Meta_score < moyenne_IMDB_Rating) :
    print("Non les critiques Metascore ne sont pas plus séveres que les utilisateurs IMDB")
else :
    print("Les critiques Metascore et les utilisateurs IMDB ont donnés les meme notes")
```

## Les films longs sont-ils mieux notés que les films courts ?

```python
notes_films_longs = movies[movies.Timing == "Long"].IMDB_Rating.mean()
notes_films_courts = movies[movies.Timing == "Court"].IMDB_Rating.mean()

if(notes_films_longs > notes_films_courts) :
    print("Oui les films long sont mieux notés que les films courts")
elif(notes_films_longs < notes_films_courts) :
    print("Non les films courts sont mieux notés ue les films long")
else :
    print("Les films courts et les films long ont les meme notes")
```

## Les réalisateurs avec beaucoup de films obtiennent-ils de meilleures notes en moyenne ?

```python
#On cré une data frame dans laquelle se trouve les réalisateurs,
#le nombre de films de chaque réalisateurs et la note moyenne 
#obtenue par chaque réalisateur pour leurs films

movies_directors_stats = movies.groupby('Director').agg(
    Nb_Films=('Director', 'size'),
    Note_Moyenne=('IMDB_Rating', 'mean')
)

movies_directors_stats.sort_values(by='Nb_Films', ascending=False)

#On cherche le nombre de réalisateurs ayant réalisé au moins 5 films
beaucoup = movies_directors_stats[movies_directors_stats.Nb_Films >= 5]

#On cherche le nombre de réalisateur ayant réalisé au plus 4 films
peu = movies_directors_stats[movies_directors_stats.Nb_Films < 5]

#Moyenne des notes moyennes de beaucoup
moyenne_beaucoup = beaucoup.Note_Moyenne.mean()

#Moyenne de peu
moyenne_peu = peu.Note_Moyenne.mean()

if(moyenne_beaucoup > moyenne_peu) :
    print("Les réalisateurs ayant réalisé le plus de film on obtenu de meilleurs notes en moyenne")
elif(moyenne_beaucoup < moyenne_peu) :
    print("Les réalisateur ayant réalisé le moins de film ont obtenu de meilleurs notes en moyenne")
else :
    print("Les réalisateurs ayant réalisé le moins de film et le plus de films ont la meme note moyenne")
```

## Certaines décennies produisent-elles davantage de films bien notés ?

```python
#On cré une fonction qui calcule la fréquence ne pourcentage
def freq(effectif, effectif_total) :
    return (effectif/effectif_total)*100

#On crée une data frame contenant les decenies, le nombre de films
#par décenies et le nombre de bons films par décenies

movies_decenies_stats = movies.groupby('Decade').agg(
    Nb_Films_Decenies=('Decade', 'size'),
    Nb_Bon_Films_Decenies=('Is_Good', 'sum')
)
movies_decenies_stats

#On cré une colone Fréquence
movies_decenies_stats['Frequences'] = freq(
    movies_decenies_stats['Nb_Bon_Films_Decenies'], 
    movies_decenies_stats['Nb_Films_Decenies'])

movies_decenies_stats.sort_values(by='Frequences', ascending=False)

#On cherche les  premieres decenies ayant produits d'avantages
#de films bien notés
movies_decenies_stats.head(5)

#Une question se pose :  Es ce qu'une decenie ayant 90% de bon film
#mais avec 10 bons films sur 11 films; est meilleure qu'une decenie
#ayant 71% de bon films mais avec 40 bon films sur 56 films?
```

## Les combinaisons de genres les plus fréquents sont-il aussi les mieux notés ?

```python
#On cherche le genre le mieux noté
best_genre = movies.groupby('Genre').IMDB_Rating.mean().idxmax()

#On cherche le genre le plus fréquent
max_genre = movies.Genre.value_counts().idxmax()

if(max_genre == best_genre) :
    print("Oui le genre le plus fréquent est aussi le mieux noté")
else :
    print("Non le genre le plus fréquent n'est pas le mieux noté")

max_genre
best_genre
```
